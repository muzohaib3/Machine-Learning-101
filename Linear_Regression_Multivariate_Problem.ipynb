```python
import numpy as np
import pandas as pd
from sklearn import linear_model

df = pd.read_csv("experience.csv")
df
```

```python
df["experience"].fillna(0)

median = df["experience"].median()
median

df["test_score"] = df["test_score"].fillna(1)

df["experience"] = df["experience"].fillna(median)
df
```

```python
regression = linear_model.LinearRegression()

regression.fit(
    df[["experience", "test_score", "interview_score"]],
    df["salary"]
)

coefficient = regression.coef_
intercept = regression.intercept_
```

```python
prediction = regression.predict([[5,7,6]])
print(f"coefficient : {coefficient} , intercept: {intercept}")
print(f"prediction for 5 years, 7 test score , 6 interview score: salary = {prediction}")
```
